```python
import json

def clean_data(data):
    #remove users with empty name using list comprehension
    data['users'] = [user for user in data['users'] if user['name'].strip()]
    
    #remove redundant data
    for user in data['users']:
        #set does not have duplicate values...therefor we convert this into set and then back to list to remove the duplicate values
        user['friends'] = list(set(user['friends']))
    
    #remove inactive users
    data['users'] = [user for user in data['users'] if user['liked_pages'] or user['friends']]
    #only take those users which have atleast one friend or liked page...if both is empty classify them as inactive user

    #remove duplicate pages using sets
    unique_pages = {}
    for page in data['pages']:
        #take id of a page as key and page(everything) as the value in dictionary
        #every key is unique in dictionary..so no duplicate values
        unique_pages[page['id']] = page
    data['pages'] = list(unique_pages.values())
    return data

#load the data
data = json.load(open("data2.json","r"))
data = clean_data(data)

#dumping the cleaned data in new json file
#indent = 4 is used to make the json file pretty using indentation
json.dump(data,open("clean_data2.json","w"),indent=4)

print("data has been cleaned successfully")
```

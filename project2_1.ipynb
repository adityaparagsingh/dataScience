```python
with open("instagram_profiles.txt", "r") as f:
    data = f.read()

lines = data.split("\n")         # split every single line
lines = [l for l in lines if l]  # remove any empty lines

# group into chunks of 10 (each profile = 10 lines)
chunks = [lines[i:i+10] for i in range(0, len(lines), 10)]

print(chunks[0])   # first profile as a list
```

```python
def parse_chunk(chunk):
    username = chunk[0]
    posts = chunk[1]
    followers = chunk[2]
    following = chunk[3]
    print("username: ", username , "\n",
          "no. of posts: ", posts,"\n",
          "followers: ",  followers,"\n",
          "following: ",following,"\n")
    
parse_chunk(chunks[0])
```

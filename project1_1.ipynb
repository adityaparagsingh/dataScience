```python
import json
```

```python
#this function will load the data from json file
def load_data(filename):
    with open(filename,"r") as f:
        data = json.load(f)
    return data
```

```python
data = load_data("data.json")
data
```

```python
type(data)
```

```python
#function to display the users and their data (freinds,liked_pages,etc)
def show_data(data):
    print("User and their connections")
    for user in data['users']:
        print(f"{user['id']} : {user['name']} is friend with {user['friends']} and liked pages are {user['liked_pages']}")
    print()
    print("Page information")
    for page in data['pages']:
        print(f"{page['id']} : {page['name']}")
```

```python
show_data(data)
```

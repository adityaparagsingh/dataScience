```python
#melt and pivot
import pandas as pd
```

```python
data = {
    'Name' : ["Aditya","Rahul","Mayank","Aranya"],
    'Math': [85, 78, 92,34],
    'Science': [90, 82, 89,65],
    'English': [88, 85, 94,76]
}
```

```python
df = pd.DataFrame(data)
```

```python
print(df)
```

```python
df2 = df.melt(id_vars = ["Name"],value_vars = ['Math','Science','English'],var_name = 'Subject',value_name='Score').copy()
```

```python
df2
```

```python
df
```

```python
df2.pivot_table(index = 'Name',columns = 'Subject', values = 'Score')
```

```python
#aggregation and grouping 
import pandas as pd
```

```python
df = pd.DataFrame({
    "Department": ["HR", "HR", "IT", "IT", "Marketing", "Marketing", "Sales", "Sales"],
    "Team": ["A", "A", "B", "B", "C", "C", "D", "D"],
    "Gender": ["M", "F", "M", "F", "M", "F", "M", "F"],
    "Salary": [85, 90, 78, 85, 92, 88, 75, 80],
    "Age": [23, 25, 30, 22, 28, 26, 21, 27],
    "JoinDate": pd.to_datetime([
        "2020-01-10", "2020-02-15", "2021-03-20", "2021-04-10",
        "2020-05-30", "2020-06-25", "2021-07-15", "2021-08-01"
    ])
})  
df
```

```python
df.groupby("Department")["Salary"].mean()
```

```python
df.groupby("Department")["Salary"].sum()
```

```python
df.groupby("Department")["Salary"].min()
```

```python
df.groupby("Department")["Salary"].max()
```

```python
df.groupby("Department")["Salary"].count()
```

```python
df.groupby("Department")["Salary"].agg(["mean","max","min"])
```

```python
df["deptAvg"] = df.groupby("Department")["Salary"].transform("mean")
```

```python
df
```

```python
df.groupby("Department").filter(lambda x:x["Salary"].mean()>85)
```

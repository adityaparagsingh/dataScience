```python
import numpy as np
import time as t

size = 1_000_000

lst1 = list(range(size))
lst2 = list(range(size))

start = t.time()
add = [x+y for x,y in zip(lst1,lst2)]  #list comprehension
end = t.time()
print(add[0:100])
print(end-start)
```

```python
size = 1_000_000

lst1 = np.array(list(range(size)))
lst2 = np.array(list(range(size)))

start = t.time()
add = lst1+lst2
end = t.time()
print(add[0:100])
print(end-start)
```

```python
print(type(add))
print(add.shape)   #shows dimension of the numpy array
```

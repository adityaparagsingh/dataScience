```python
#people you may know
import json

def load_data(filename):
    with open(filename,"r") as f:
        return json.load(f)  #load json data from file f

def find_people_you_may_know(user_id,data):
    user_friends = {}
    
    for user in data['users']:
        #creating unique friends dictionary
        user_friends[user['id']] = set(user['friends'])
        user_friends[user['name']] = set(user['friends'])

    if user_id not in user_friends:
        return[]  #if invalid user (id) then return empty people u may kmow dict or return any msg (data sanitisation)

    direct_friends = user_friends[user_id]

    suggestion = {}
    
    for friend in direct_friends: #taking all the friends one by one
        for mutual in user_friends[friend]: #if friend is present in user_friend..it is a mutual friend
            if mutual!=user_id and mutual not in direct_friends:
                #count mutual friends
                #if a user have more count of mutual freinds with user 2 then it will be recommended more
                suggestion[mutual] = suggestion.get(mutual,0) + 1 #if mutual friend is not present in suggestion dict ,,return 0 or else do +1...if another mutual is present between 2 users again do +1
    
    sorted_suggestion = sorted(suggestion.items(),key = lambda x : x[1],reverse = True )  #sort on the basis of more mutual friends
    #.item converts each key value pair into a tuple and converts dict into list of tuples
    #lambda is used to create function without a name ..single line function for instant use on iterative data
    #reverse = True will sort in descending order...more mutuals more priority

    return (user_id for user_id, _ in sorted_suggestion)  #_ represent the second value in tuple i.e. count of mutual friends ..used to ignore the second value

data = load_data("sample_massive_data.json")
user_id = 3
recommendation = find_people_you_may_know(user_id,data)
print(list(recommendation))
```
